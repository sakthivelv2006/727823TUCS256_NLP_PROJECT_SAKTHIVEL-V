```python
import pandas as pd
import re
import spacy
from nltk.corpus import stopwords
from textblob import TextBlob

# Load only required resources
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])
stop_words = set(stopwords.words("english"))

# Read CSV
df = pd.read_csv("customer_support_tickets.csv")

TEXT_COLUMN = "Ticket Description"

def process_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    corrected = str(TextBlob(text))
    
    doc = nlp(corrected)
    
    tokens = [t.text for t in doc if t.is_alpha and t.text not in stop_words]
    lemmas = [t.lemma_ for t in doc if t.is_alpha and not t.is_stop]
    
    if "account" in corrected or "login" in corrected:
        category = "Account Issue"
    elif "payment" in corrected or "refund" in corrected:
        category = "Payment Issue"
    elif "order" in corrected or "delivery" in corrected:
        category = "Order Issue"
    elif "error" in corrected or "technical" in corrected:
        category = "Technical Issue"
    else:
        category = "General Query"
    
    return pd.Series([corrected, tokens, lemmas, category])



df[["Processed_Text", "Tokens", "Lemmas", "Category"]] = df[TEXT_COLUMN].apply(process_text)


print(df[["Processed_Text", "Tokens", "Lemmas", "Category"]].head(3))
```
